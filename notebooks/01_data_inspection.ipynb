Computation of the average time between 2 measurments:

```python
time_stamp= sample.End_Time.drop_duplicates()  # pick up the time values
time_stamp=time_stamp.map(lambda x:pd.to_datetime(x,format="%Y%m%d%H%M%S")).sort_values() # conversion in dates, sorted out
#print(time_stamp)
time_res=pd.DataFrame([time_stamp.iloc[i]-time_stamp.iloc[i-1] for i in range( 1,len(time_stamp))]).mean() #computing the means (NB: non cyclic computing)
print("the average gap of time is of "+ str(time_res.iloc[0]) + " for this sample")
```
